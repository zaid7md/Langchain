```python
# BaseModel is the core class in Pydantic.
# It is used to define structured data models using Python type hints.

# BaseModel automatically:
# - validates incoming data
# - converts data to the correct types
# - raises clear errors if data is invalid
# - creates clean Python objects from raw input

# It acts as a strict boundary between external/untrusted data
# (APIs, JSON, env vars, LLM outputs)
# and your internal Python code.

# In short:
# BaseModel = "declare data shape + get validation for free"


# Field is a helper function in Pydantic used to add metadata,
# constraints, and default behavior to model fields.

# It allows you to:
# - set default values
# - define validation rules (min/max, regex, length, etc.)
# - add descriptions and examples
# - control serialization behavior

# Field does NOT create a field by itself;
# it enhances a field defined via a type hint.

# Example mental model:
# type hint  → "what type this is"
# Field(...) → "rules and metadata for this field"

# Common uses of Field:
# - value constraints (gt, lt, ge, le)
# - string rules (min_length, max_length, regex)
# - defaults and default_factory
# - documentation (description, example)

# Field is heavily used in:
# - API schemas (FastAPI)
# - tool input validation (LangChain)
# - structured LLM outputs

# In short:
# Field = "attach rules, defaults, and metadata to a model field"
    
```

```python
from pydantic import BaseModel , Field #Pydantic is a data validation and parsing library for Python.
```

```python
class User(BaseModel):
    # 'id' must be a positive integer
    id: int = Field(gt=0, description="Unique user ID")

    # 'name' must be between 3 and 50 characters
    name: str = Field(min_length=3, max_length=50)

    # 'age' must be between 18 and 100
    age: int = Field(ge=18, le=100)

    # default value if not provided
    is_active: bool = Field(default=True)


user1 = User(id = "123" , name = "John Doe" , age = "32" ,  is_active = True)
#this works becuase of the pydantic model 
```
