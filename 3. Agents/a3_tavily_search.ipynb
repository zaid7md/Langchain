```python
import os 
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain.agents import create_agent
from langchain.tools import tool
from langchain_core.messages import HumanMessage
# from tavily import TavilyClient 
from langchain_tavily import TavilySearch #tavily provides a tool that can be used to search the web

load_dotenv(override=True)

api_key = os.getenv("LLM_API_KEY")
tavily_api_key = os.getenv("TAVILY_API_KEY")

if api_key and tavily_api_key:
    print("Api keys loaded successfully")
else:
    print("Api keys not found")

```

```python
llm = ChatOpenAI(model="openai/gpt-4o-mini", temperature=0 , api_key=os.getenv("LLM_API_KEY") , base_url=os.getenv("BASE_URL"))
tools = [TavilySearch()]
agent = create_agent(llm, tools)
```

```python
result = agent.invoke({"messages" : HumanMessage(content="give me top5 best football players in the world")})
print(result)
```
