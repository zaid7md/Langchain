```python
#by default langchain has tavily for web search
# Tavily is a web search API designed specifically for LLMs
# It allows an LLM to fetch real-time, up-to-date information from the internet

# LLMs by default:
# - Cannot browse the web
# - Have a knowledge cutoff
# - May hallucinate when asked about current information

# Tavily solves this by:
# - Performing live web searches
# - Returning clean, structured, LLM-friendly results
# - Providing summaries instead of raw HTML
# - Including relevant sources and context

# In LangChain, Tavily is commonly used as:
# - A Tool (for agents)
# - A Retriever (for web-based RAG)

# Typical flow when using Tavily with LangChain:
# 1. User asks a question
# 2. LLM decides it needs current or external information
# 3. LangChain agent calls Tavily search
# 4. Tavily returns relevant web results
# 5. LLM reads the results and generates a grounded answer

# Tavily is useful for:
# - Real-time knowledge (news, versions, pricing, policies)
# - Reducing hallucinations
# - AI agents that reason and act
# - Research assistants and chatbots
# - Web-based RAG without managing your own vector database

# Mental model:
# - LangChain = orchestration (reasoning, tools, memory)
# - Tavily = internet access for the LLM

# In short:
# Tavily integrates live web search into an LLM pipeline
```
