```python
from dotenv import load_dotenv
import os 
from langchain_openai import ChatOpenAI
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.messages import HumanMessage
from langchain_tavily import TavilySearch
from langchain.agents import create_agent
from langchain.tools import tool
from pydantic import BaseModel , Field 


load_dotenv(override='True')

api_key = os.getenv("LLM_API_KEY")
if not api_key:
    print("LLM_API_KEY not found")
else:
    print("Api key exists")
```

```python
class Source(BaseModel):
    url : str = Field("url of the source")

class AgentResponse(BaseModel):
    answer : str = Field(description= "Agents answer to the query")
    sources : list[Source] = Field("List of sources used to answer the query")
```

```python
llm = ChatOpenAI(model="gpt-4o", temperature=0)
tools = [TavilySearch()]
agent = create_agent(llm, tools , response_format=AgentResponse)
#response_format tells the agent what shape the final answer must have.
#instead of a string, the answer will be a pydantic model , json or any other data structure.
#this is useful when you want to return a structured answer.
```

```python
result = agent.invoke({"messages" : HumanMessage(content="Give me 5 jobs in the filed of langchain")})
```

```python
print(result["structured_response"].answer)
#now there is a structured answer in the output which you can view in the langchain tracing 
#also langchain agents or llms answer in dict form , so you need to access the structured_response to get the actual answer.
```

```python
print(result["structured_response"].sources)
```
