```python
from dotenv import load_dotenv
import os 

load_dotenv(override=True)
api_key =  os.getenv("LLM_API_KEY")
base_url = os.getenv("BASE_URL")

if(api_key and base_url):
    print("Api key and base url found")
else:
    print("Api key and base url not found")
```

```python
from langchain_openai import ChatOpenAI
from langchain.agents import create_agent 
from langchain_tavily import TavilySearch
```

```python
from typing import List
from pydantic import BaseModel , Field

class Source(BaseModel):
    """Schema for a source used by the agent"""

    url : str = Field(description="The url of the source")

class AgentResponse(BaseModel):
    """Scehama for agent response with answer and sources"""

    answer : str = Field(description="Agent's answer to query")
    sources : List[Source] = Field(
        default_factory=list , description="List of sources used by the agent to answer the query"
    )
```

```python
model = ChatOpenAI(model="openai/gpt-4o", temperature=0 , api_key=os.getenv("LLM_API_KEY") , base_url=os.getenv("BASE_URL"))
tools = [TavilySearch()]
```

```python
agent = create_agent(
    model, 
    tools, 
    response_format=AgentResponse,
)
```

```python
result = agent.invoke({
    "messages":[
        {
            "role" : "user",
            "content" : "Search for 3 job postings as an ai engineer using langchain in india"
        }
    ]
}
)
```
