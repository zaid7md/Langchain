```python
from dotenv import load_dotenv
import os 
from langchain_openai import ChatOpenAI
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.messages import HumanMessage
from langchain_tavily import TavilySearch
from langchain.tools import tool
from pydantic import BaseModel , Field 
from langchain_classic.agents.react.agent import create_react_agent
from langchain_classic.agents import AgentExecutor
from langchain_classic import hub #Think of it like GitHub, but specifically for LangChain prompts and chains.
#so this is a place where you can find structured prompts and chains 
# load_dotenv(override='True')

api_key = os.getenv("LLM_API_KEY")
if not api_key:
    print("LLM_API_KEY not found")
else:
    print("Api key exists")
```

```python
tools = [TavilySearch()]
llm = ChatOpenAI(model="openai/gpt-4o", temperature=0 , api_key=os.getenv("LLM_API_KEY") , base_url=os.getenv("BASE_URL"))
react_prompt = hub.pull('hwchase17/react')
```

```python
print(react_prompt.pretty_print())
```

```python
agent = create_react_agent(
    llm, 
    tools,  
    react_prompt)
```

```python
agent_executor = AgentExecutor(agent = agent , tools = tools , verbose = True , handle_parsing_errors=True
)
```

```python
chain  = agent_executor
```

```python
result =  chain.invoke(
    input = {
        "input" : "Search for 3 job posting for an ai engineer using langchain in the bay area on linkedin and list theri details"
    }
)
```

```python
# ReAct agent loop (clean, correct version)

# Here’s the flow in plain English, mapped to what actually happens:

# 1️⃣ User input arrives

# The agent receives the user’s question.

# 2️⃣ LLM reasons about the input

# It decides whether it needs a tool.

# This “thinking” is implicit (even if shown as Thought: text).

# 3️⃣ LLM requests a tool

# It chooses:

# which tool

# what input to pass

# This is a request, not execution.

# 4️⃣ LangChain executes the tool

# The LLM pauses.

# LangChain runs the tool function in Python.

# 5️⃣ Tool output is fed back

# The result is appended as an Observation.

# 6️⃣ LLM reasons again

# Now it has new information.

# It may:

# answer

# request another tool

# refine the request

# 7️⃣ Loop continues

# Until:

# the LLM outputs Final Answer

# OR max iterations/time is hit

# ✔️ That’s ReAct.
```
