TypedDict


```python
from typing import TypedDict, List
#TypedDict lets you define the expected “shape” of a dictionary using types.
```

```python
class State(TypedDict):
    messages: List[str]
    step: int
```

```python
# TypedDict is used to define the expected "shape" of a dictionary
# Think of it like a TypeScript interface, but for Python dicts

# ---------------------------------------------------------
# Why TypedDict exists
# ---------------------------------------------------------
# Normal Python dictionaries are unstructured:
#   - Any key can exist
#   - Any value type is allowed
#   - Errors only show up at runtime
#
# TypedDict adds:
#   - Fixed keys
#   - Expected value types
#   - IDE autocomplete
#   - Static type checking (mypy, pyright)

# ---------------------------------------------------------
# JavaScript / TypeScript analogy
# ---------------------------------------------------------
# TypeScript:
# interface State {
#   messages: string[]
#   step: number
# }

# Python equivalent using TypedDict:
#
# from typing import TypedDict, List
#
# class State(TypedDict):
#     messages: List[str]
#     step: int

# ---------------------------------------------------------
# What TypedDict DOES NOT do
# ---------------------------------------------------------
# - It does NOT validate data at runtime
# - Python will NOT throw errors automatically
# - Wrong types can still exist unless checked manually
#
# Example:
# state = {"step": "one"}   # This is allowed at runtime

# TypedDict is for:
# - Type safety
# - Better documentation
# - Fewer bugs during development

# ---------------------------------------------------------
# Why LangGraph uses TypedDict
# ---------------------------------------------------------
# LangGraph passes a shared "state" between nodes
# Each node reads and updates part of the state
#
# TypedDict clearly defines:
# - What keys exist in state
# - What data type each key holds
# - What nodes are allowed to read/write

# ---------------------------------------------------------
# LangGraph-style example
# ---------------------------------------------------------
# from typing import TypedDict, List
#
# class AgentState(TypedDict):
#     messages: List[str]
#     next: str
#
# def agent_node(state: AgentState):
#     last_message = state["messages"][-1]
#     return {
#         "messages": state["messages"] + ["thinking..."]
#     }

# ---------------------------------------------------------
# TypedDict vs Pydantic
# ---------------------------------------------------------
# TypedDict:
# - Static typing only
# - No runtime validation
# - Very lightweight
# - Ideal for internal state (LangGraph)
#
# Pydantic:
# - Runtime validation
# - Used for APIs and user input
# - Adds overhead
# - Ideal for FastAPI request/response models

# ---------------------------------------------------------
# Rule of thumb
# ---------------------------------------------------------
# - Use TypedDict for LangGraph state
# - Use Pydantic for FastAPI APIs
# - Do NOT use TypedDict for API validation
```

```python
#union means : “This variable can be one of multiple types.”
```

```python
from typing import Union

x: Union[int, str]
y : int 
```

```python
x=  12 ; 
x=  "ddsds" #x can be int or str 
```

```python
def parse_id(value: Union[int, str]) -> int:
    if isinstance(value, int):
        return value
    return int(value)
```

```python

class ToolResult(TypedDict):
    output: Union[str, dict]


class AgentState(TypedDict):
    result: Union[str, list[str]]
```
