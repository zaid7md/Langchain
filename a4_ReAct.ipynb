```python
# ReAct = Reason + Act
# A ReAct-style agent is an LLM that can:
# 1. Reason step-by-step about a problem
# 2. Decide when to take an action (use a tool)
# 3. Observe the result of that action
# 4. Repeat this loop until it reaches a final answer

# Core ReAct loop:
# Thought  -> Reasoning about what to do next
# Action   -> Calling a tool (search, calculator, API, DB, etc.)
# Observation -> Result returned by the tool
# Repeat until the task is solved

# Example reasoning flow:
# Thought: I need some external information
# Action: Call a search tool
# Observation: Tool returns data
# Thought: Now I can compute or decide
# Final Answer: Respond to the user

# Why ReAct agents are powerful:
# - They reduce hallucinations
# - They can use real-world tools
# - They can solve multi-step problems
# - They make dynamic decisions at runtime

# Difference between a normal chain and a ReAct agent:
# - Chain: Fixed sequence of steps, no decisions
# - ReAct Agent: Decides actions dynamically based on reasoning

# ReAct vs Function Calling:
# - Function calling: LLM calls one function, logic handled by developer
# - ReAct agent: LLM controls the full workflow and tool usage

# In LangChain:
# - ReAct behavior is built into agents
# - The LLM decides when and which tool to call
# - You only define the tools and the model

# Mental model:
# Function calling = assistant
# ReAct agent = junior engineer that thinks and acts
```
